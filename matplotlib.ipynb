```python
import matplotlib.pyplot as plt
RK=[8,7,13,24,32,50]
tiert1=["IIT dhanbad(jharkhand)","IIT roorkel(uttrakhand)","IIT madras(chennai)","IIT kharagpal(westbengal)","IIT guwahat(assam)","IIT hyderabad(telangana)"]
tiert2=["IIT DM(kurnool)","IIT kalyani(west bengal)","IIT luknow","IIT sricity(AP)","IIT hyderabad(telangana)","IIT ongole(AP)"]
plt.plot(tiert1,RK,marker="D",label="graph")
plt.legend()
plt.plot(tiert2,RK,label="IIt colleges",linestyle="-.",color="green")
plt.legend()
plt.xticks(rotation=45)
plt.title("NIRF OF IIT AND IIIT colleges")
plt.xlabel("college positions")
plt.ylabel("college rankings")
plt.show()
```

```python
import matplotlib.pyplot as plt
w=["week1","week2","week3","week4","week5"]
temp=[20,40,80,60,43]
plt.bar(w,temp)
plt.title("weekly temperature")
plt.xlabel("week")
plt.ylabel("temp")
plt.show()

```

```python
import matplotlib.pyplot as plt
i=["bag","pens","dress","chains","bible"]
p=[500,40,800,60,400]
#plt.barh(i,p,color=["pink","green","orange","red","black"])
plt.scatter(i,p,color=["pink","green","orange","red","black"])
plt.xlabel("items")
plt.ylabel("price")
plt.title("items")
plt.show()
```

```python
import matplotlib.pyplot as plt
items=["crab","prawns","fish","octopus","starfish","chicken","mutton"]
price=[350,200,300,800,700,320,1000]
plt.pie(price,labels=items,autopct="%1.1f%%")
plt.title("weekend purchase",color="red")
plt.show()

```
