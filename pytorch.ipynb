```python
#indexing
h=torch.tensor([[10,12,13],[14,15,16],[17,18,19]])
print(h[1])
print(h[2])
print(h[[1,2]])
print(h[[1,2],1])
print(h[[0,1,2]])
print(h[[0,1,2],2])
```

```python
#slicing
print(h[0:3,0:1])
print(h[:3,0])
print(h[:,0])
print(h[:,0:1])
```

```python
print(h[2:3,0:3])
print(h[2:3,:3])
print(h[2:,:3])
print(h[2:3,:])
```

```python
f=torch.tensor([[2,8,10],[4,5,6],[10,11,12]])
print(torch.sum(f))
print(f.sum())
```

```python
l=torch.tensor([[2,8,10],[4,5,6],[10,11,12]])
m=torch.tensor([[10,12,13],[14,15,16],[17,18,19]])
print(torch.subtract(l,m))
print(torch.mul(l,m))
print(torch.divide(l,m))
```

```python
print(l.max())
print(l.min())
print(l.argmax())
```
