```python
c=pd.Series([6,6,10,4,5,6],name="apples")
print(c)
print(c.mean())
print(c.max())
print(c.min())
print(c.unique())
print(c.values)
print(c.size)
print(c.dtype)
print(c.index)
print(c.name)

```

```python
import pandas as pd#file conversion
a=pd.DataFrame({"sports":["cricket","vollyball","carrom","rugby","batminton"],"Ranking":[1,6,8,5,25],"preference":[6,8,7,1,2],"venue":["VSKP","YSR","NLR","KDP","ELR"]})
print(a)
w=a.to_csv("sports.csv")
print(w)
x=a.to_excel("sports.xlsx")
print(x)
z=a.to_json("sports.json")
print(z)

```
