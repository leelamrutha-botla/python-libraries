```python
j=tf.constant([[1,2,3],[3,5,6],[4,5,6]])#indexing
print(j[1,1])
print(j[1][1])
print(j[1])
print(j[2])
h=tf.transpose(j)
print(h[0])
i=tf.constant([[0,1,5,6],[1,5,4,3],[1,2,4,6]])
print(i)
k=tf.transpose(i)
print(k[2])
#indexing
print(k[2:3,0:4])
print(k[2:3])
print(i[2:4])
print(i[1:2,2:3])
print(i[1:3,2:4])
print(i[0:2,1:3])
print(i[-1,-3])
print(i[-1:-2:-1,-3:-4:-1])
print(i[-1:,-3])
l=tf.constant([2,6,8,0,3])
print(tf.reverse(l,axis=[0]))
```

```python
#slicing
l=tf.constant([[2,6,8,0,3],[4,5,7,8,6]])
print(tf.reverse(l,axis=[0]))
print(tf.reverse(l,axis=[1]))
print(tf.reverse(l,axis=[1,0]))
print(tf.reverse(l,axis=[0,1]))
p=tf.constant([2,6,8,0])
q=tf.constant([9,5,7,4])
b=tf.reshape(p,(1,4))
print(b)
print(p.shape)
o=tf.reshape(p,(1,-1))
print(o)
x=tf.reshape(l,[-1])
print(x)
m=tf.add(p,q)
print(m)
print(tf.subtract(p,q))
print(tf.multiply(p,q))
print(tf.divide(p,q))
```

```python
import tensorflow as tf
b=tf.constant([[3,4],[5,6],[7,8]])
print(tf.reduce_mean(b))
print(tf.reduce_sum(b))
print(tf.reduce_max(b))
print(tf.reduce_min(b))
print(tf.reduce_mean(b,axis=0))
print(tf.reduce_sum(b,axis=0))
print(tf.reduce_max(b,axis=0))
print(tf.reduce_min(b,axis=0))
print(tf.reduce_mean(b,axis=1))
print(tf.reduce_sum(b,axis=1))
print(tf.reduce_max(b,axis=1))
print(tf.reduce_min(b,axis=1))
```
