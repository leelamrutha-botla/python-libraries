```python
import torch
a=torch.tensor([6,18,24])
b=torch.tensor([[10,15,18],[20,21,22]])
print(a)
print(b)
print(b.shape)#attribute
print(b.size())#funtion
print(b.ndim)#dimension
c=torch.tensor([99.,96,54])
print(c)
print(c.dtype)
d=torch.full((3,5),7)
print(d)
e=torch.empty(6)
print(e)
f=torch.zeros(1,4)
print(f)
g=torch.ones(1,2)
print(g)
h=torch.range(96,108)
print(h)
i=torch.range(96,108,6)
print(i)
j=torch.eye(3)
print(j)
```

```python
import torch
a=torch.tensor([[6,15,21],[4,3,1]])
print(a)
print(a[1,1])#indexing
print(a[1:2,1:2])#slicing
print(a[1:3,2:3])
#concatenation
b=torch.tensor([[1,2],[3,4],[5,6]])
c=torch.tensor([[18,7],[8,9],[10,11]])
print(torch.cat((b,c),dim=1))#column concatenation
print(torch.cat((b,c),dim=0))#row concatenation
d=torch.tensor([[[2,3,4],[5,6,7],[1,2,4]]])
e=torch.tensor([[[6,7,8],[8,6,9],[0,7,8]]])
print(torch.cat((d,e),dim=1))
print(torch.cat((d,e),dim=0))
f=torch.tensor([[1,9],[2,5],[4,6]])
g=torch.tensor([[1,8,9],[5,7,6],[6,8,6]])
print(torch.cat((f,g),dim=1))
#print(torch.cat((f,g),dim=0))
h=torch.tensor([[1,9],[2,5],[4,6]])
i=torch.tensor([[1,8],[5,6]])
print(torch.cat((h,i),dim=0))
#splitting
j=torch.tensor([1,10,11,12,15,16])
print(torch.split(j,2))
k=torch.arange(20)
print(torch.split(k,4))
print(torch.split(k,[3,7,5,4,1]))
#chunk
print(torch.chunk(h,3))
#zeros
l=torch.zeros(4,5,6)
print(l)
print(l.size(1))
m=torch.ones(4,5,6)
print(m)
print(m.size(2))
```
