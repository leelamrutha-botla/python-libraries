```python
import numpy as np
#creating 1D,2D,3D arrays
a=np.array([10,4,5])
print(a)
b=np.array([[12,4,5]])
print(b)
c=np.array([[[2,4,6]]])
print(c)
d=np.array([[1,4,6],[4,6,7]])
print(d)
e=np.array([[[10]]])
print(e)
f=np.array([[[[22,4,5],[4,5,6],[6,6,7]]]])
print(f)

```

```python
a=np.array(['25','5.6',"hi",True])
print(a)
n=np.array([[2,3,4],[6,6,7]])
print(n)
g=np.zeros(6)
print(g)
k=np.zeros((6,5),dtype=int)
print(k)
l=np.ones(3)
print(l)
p=np.ones((2,1),dtype=int)
print(p)
d=np.empty((2,1),dtype=int)
print(d)
s=np.arange(4)
print(s)
t=np.arange(1,6,2)
print(t)
```

```python
import numpy as np
a=np.eye(3,4,k=2)
print(a)
i=np.identity(4)
print(i)
w=np.full((4,5),'ammu')
print(w)
l=np.random.seed(10)
print(l)
u=np.random.rand(4,3)
print(u)
o=np.random.randint(3,7,(4,5))
print(o)
e=np.array([[2,3,4],[4,5,6]])
print(e)
print(e.shape)
print(e.size)
print(e.ndim)
print(e.dtype)
print(e.itemsize)
```
